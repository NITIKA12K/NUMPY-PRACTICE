**ARRAY OPERATIONS**

```python
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)
print(a1)
a2
```

```python
#scalar operations     on one array
#arithmetic
print(a1**2)
#relational
a2==14
```

```python
#vector operations               on two arrays
#arithmetic
a1+a2
```

**ARRAY FUNCTIONS**

```python
a1=np.random.random((3,3))
a1=np.round(a1*100)
a1
```

```python
#max/min/sum/product
#0-->column,1-->row
np.max(a1,axis=0)
np.min(a1)
np.sum(a1)
np.prod(a1)
```

```python
#mean/median/mode/std/var
np.mean(a1,axis=1)
```

```python
#trigonometric functions
np.sin(a1)
```

```python
#dot product
a2=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)
print(a2)
print(a3)
np.dot(a2,a3)
```

```python
#log and exponents
np.exp(a2)
np.log(a2)
```

```python
#round,floor,ceil
print(np.random.random((2,3)))
np.ceil(np.random.random((2,3))*100)
```

**INDEXING AND SLICING**

```python
a1=np.arange(10) #1D
a2=np.arange(12).reshape(3,4)   #2D
a3=np.arange(8).reshape(2,2,2) #3D
print(a1)
a1[-1]
```

```python
print(a2)
a2[2,3]
```

```python
print(a3)
a3[1,0,1]
```

```python
#slicing
print(a1)
a1[2:5]
a1[2:5:2]
```

```python
print(a2)
a2[0,:]
a2[:,2]
a2[1:,1:3]
a2[::2,::3]
a2[0::2,1::2]
a2[1,::3]
a2[:2,1:]
```

```python
a3=np.arange(27).reshape(3,3,3)
a3
```

```python
a3[2]
a3[::2]
a3[0,1,:]
a3[1,0:,1]
a3[2,1:,1:]
a3[::2,0,::2]
```

**STACKING-HORIZONTAL AND VERTICAL**

```python
a4=np.arange(12).reshape(3,4)          #horizontal stacking = same rows
a5=np.arange(12,24).reshape(3,4)
np.hstack((a4,a5))
```

```python
np.vstack((a4,a5))          #vertical stacking = same column
```
