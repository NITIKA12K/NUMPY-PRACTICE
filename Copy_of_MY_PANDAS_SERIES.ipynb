# **IMPORTING PANDAS**

```python
import numpy as np
import pandas as pd
```

# **Series from lists**

```python
#string
country=["india",'pakistan','newyork','malaysia']
pd.Series(country)           # CAPITAL S
```

```python
#integers
runs=[87,88,00,22,22]
runs_ser=pd.Series(runs)
```

```python
#custom index + #Setting a name
marks=[89,67,45,78,98]
subjects=['maths','hindi','english','DBMS','GMGJ']
marks=pd.Series(marks,index=subjects,name='nitika ke marks')
marks
```

# **Series from dict**

```python
marks={                               #in dictonary you don't have to mention that this is  your index
    'english':56,
    'maths' :88,
    'hghjgj' :77,
    'oopws':88
}
marks_series=pd.Series(marks,name='nitika ke marks')
marks_series
```
